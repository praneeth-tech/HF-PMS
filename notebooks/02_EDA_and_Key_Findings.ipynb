```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_excel('/content/Pharma_Patient_Dataset_60_Records.xlsx')

plt.hist(df['Age'], bins=10)

plt.title('Age Distribution')

plt.xlabel('Age')

plt.ylabel('Patients')

plt.show()
```

```python
df['Disease'].value_counts().plot(kind='bar')

plt.title('Disease Distribution')

plt.show()
```

```python
df['Drug/Name'].value_counts().plot(kind='bar')

plt.title('Drug Usage')

plt.show()
```

```python
df['Outcome'].value_counts().plot(kind='pie', autopct='%1.1f%%')

plt.title('Treatment Outcome Distribution')

plt.show()
```
