```python
import pandas as pd

df = pd.read_csv('/content/diabetic_data.csv')

print(df.shape)
df.head()
```

```python
total_patients = len(df)

print("Total Patients:", total_patients)
```

```python
gender_count = df['gender'].value_counts()

print(gender_count)
```

```python
age_distribution = df['age'].value_counts()

print(age_distribution)
```

```python
readmission = df['readmitted'].value_counts()

print(readmission)
```

```python
avg_hospital_days = df['time_in_hospital'].mean()

print("Average Time in Hospital:", avg_hospital_days)
```

```python
avg_lab = df['num_lab_procedures'].mean()

print("Average Lab Procedures:", avg_lab)
```

```python
avg_medications = df['num_medications'].mean()

print("Average Medications:", avg_medications)
```

```python
avg_diagnoses = df['number_diagnoses'].mean()

print("Average Diagnoses:", avg_diagnoses)
```

```python
admission_type = df['admission_type_id'].value_counts()

print(admission_type)
```
